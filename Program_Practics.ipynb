```python
#Palindrom
i = input("Enter a value: ")
reverse_input = i[::-1]

if i == reverse_input:
    print("The input is a palindrome.")
else:
    print("The input is not a palindrome.")

```

```python
#anagram
a = 'silent'
b = 'listen'
result = False
if len(a) == len(b):
  for i in a:
    if i not in b:
      break
    else:
      result = True
result
```

```python
#factorial
def fact(num):
  if num == 0 or num == 1:
    return 1
  return num * fact(num - 1)

fact(3)
```

```python
#* pattern
for i in range(1,6,1):
  for j in range(i):
    print('* ',end='')
  print()
```

```python
for i in range(3):
  for j in range(3):
    if i == j:
      print('1 ',end='')
    else:
      print('0 ',end='')
  print()
```

```python
#Check whether a number is prime.
num = 37
isPrime = True
for i in range(2,num):
  if(num % i == 0):
    isPrime = False
isPrime
```

```python
#Print all prime numbers between 1 and 100.
for i in range(2,101):
  isPrime = True
  for j in range(2,i):
    if i % j == 0:
      isPrime = False
      break
  if isPrime:
    print(i)
```

```python
#Find duplicate elements in a list.
l = [3,6,6,7,4,3]
s = set(l)

for i in s:
  if l.count(i) > 1:
    print(i)
```

```python
#Find the first non-repeating character.
a = 'hello'

for i in a:
  if a.count(i) == 1:
    print(i)
    break
```

```python
#Find the longest word in a sentence.
s = 'hello welcome to python programming'
l = list(s.split(' '))
longestLength = 0
longestWord = '1'
for i in l:

  if len(i) > longestLength:
    longestLength = len(i)
    longestWord = i
longestWord
```

```python
#Longest substring without repeating characters
s = "abcabcdbb"
substring = ''
longestSubstring = ''
for i in s:
  if i not in substring:
    substring += i
  else:
      if len(substring) > len(longestSubstring):
        longestSubstring = substring
longestSubstring
```
