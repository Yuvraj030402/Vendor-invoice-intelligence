```python
#Palindrom
i = input("Enter a value: ")
reverse_input = i[::-1]

if i == reverse_input:
    print("The input is a palindrome.")
else:
    print("The input is not a palindrome.")

```

```python
#anagram
a = 'silent'
b = 'listen'
result = False
if len(a) == len(b):
  for i in a:
    if i not in b:
      break
    else:
      result = True
result
```

```python
#factorial
def fact(num):
  if num == 0 or num == 1:
    return 1
  return num * fact(num - 1)

fact(3)
```

```python
#* pattern
for i in range(1,6,1):
  for j in range(i):
    print('* ',end='')
  print()
```

```python
for i in range(3):
  for j in range(3):
    if i == j:
      print('1 ',end='')
    else:
      print('0 ',end='')
  print()
```
